### Plotly makes these visual interfaces a lot easier and the overall graphing interface a lot easier. When you're able to hover over it, it gives you values, and you're able to play around with it. You get a lot more scope to know what actual values there are. 

```python
df[['A','B']].plot(kind='bar') #Also able to do it with two columns 
```

You can call these using either df.plot(kind='...') or specific accessors like df.plot.bar():
• scatter (df.plot.scatter() or df.plot(kind='scatter'))
• line (df.plot.line() or df.plot(kind='line'))
• area (df.plot.area() or df.plot(kind='area'))
• bar (df.plot.bar() or df.plot(kind='bar'))
• barh (df.plot.barh() or df.plot(kind='barh'))
• hist (df.plot.hist() or df.plot(kind='hist'))
• box (df.plot.box() or df.plot(kind='box'))
