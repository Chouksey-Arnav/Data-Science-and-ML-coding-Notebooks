```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
#Run all necessary imports
```

```python
df = pd.DataFrame({'col1':[1,2,3,4],'col2':[444,555,666,444],'col3':['abc','def','ghi','xyz']})
df.head() #Create a DataFrame
```

##  Now let's work on finding unique values in a pandas DataFrame

```python
#There is 3 methods for finding unique values.
```

Lets say, you wanted to find all the unique values in column 2 of our pandas DataFrame. Here's how to do that. 


```python
df['col2'].unique() #All you need is a .unique() method in pandas and it will return all the unique values!
```

## But let's say, instead of wanting an array of numbers that are unique in any given column, you want the number of unique numbers in any given column. 

```python
#You do....
df['col2'].nunique()
```

### Now, if you want a table of all the unique values and how many times they show up, you can use the `value_counts` function in pandas. 

```python
df['col2'].value_counts()
#Gives you a table of the values in the dataframe and how many of each value there is in the specific column
```

### Next, let's transfer into conditional selection and see the power of that... 

```python
#Lets find where col1 is greater than col2
df[df['col1']>df['col2']]
#Doesnt return anything because its not true any time...
```

```python
#BUT Lets add a 'pipe' (OR) operator and see what happens now...
df[(df['col1'] > df['col2']) | (df['col1'] > 2)]
```

```python
df #Here is he dataframe again
```

### Now let's talk about using the `.apply` function in pandas. The use case of this is, say, if I define a function and I want to apply that to a pandas DataFrame, how do I do that? That's where `.apply` comes into play. 

```python
#Using .apply, lets first define a specific funtion first
def power2(x):
    return(x**2) #Very simple. All this does is, whatever number you pass into it, it just lets out the number to the power of two. 
```

But...
How do we apply this to a pandas DataFrame? We're going to use the `.apply` method. 

```python
df['col2'].apply(power2) #Makes everything in the second column raise to the power of 2. 
```

### But now, what if we just wanted to use this expression once instead of defining a whole new function for it? We can just use the lambda expression in pandas for this. 

```python
df['col2'].apply(lambda memes: memes**2)
```

### Again, to remove columns, all we have to do is just use the `df.drop` method, and then we can name our parameter. We name what column we want to drop, and then we want to name the axis, whether that's vertical or horizontal. 

```python
df.drop('col1', axis=1, inplace=False) #What this is essentially saying is it's going to drop column 1. It's defining the axis so there's no confusion, and it's making the change permanent within `inplace = true`. 
```

```python
df #Now, it shows the updated, dropped dataframe!
```

### To find null values in your data frame, you can use the `df.isnull` method in pandas. It will return you boolean values: values that are null are `true`, and values that aren't are `false`. In our case, if we do it, our data frame will get all `false` values because we have no null values. 

```python
df.isnull() #As expected output...
```
