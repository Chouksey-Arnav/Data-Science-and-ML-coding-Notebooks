We can also remove either columns or rows in pandas by using the `df.drop` method. 

```python
#To remove a Column (Up to down)
df.drop('new', axis=1, )
```

```python
#To remove a row (Left to Right)
df.drop('A', axis=0)
df
```

The difference between axis=0 and axis=1 in the df.drop method (and in future Pandas methods) is telling us whether to remove the column or the row.
axis=0 tells it to remove a row (left to right), while axis=1 tells us to remove a column (top to bottom)

```python
df
```

A LOT of pandas methods also have a inplace arguement which tells is if we want a arguement to actually alter the dataFrame so when we use it in the furure the dataframe is actually altered 

For example, most pandas methods that use this in-place argument are set to `False`. If we add it so that `inplace` is `True` after the `axis` argument, then if I take out something with the `df.drop` function, for example, like `new`, it stays that way.

In a normal one, without putting `inplace` to be `True` or `False`, it might still show that `drop` function. When I run that specific drop function. The new column won't show because I've dropped it for that specific time, but then, when I rerun `df` in a whole new cell, it will still show up with the new column (because `in place` is automatically preset to `false` unless I reinstate it otherwise). 

```python
df # Even though we have 'dropped' columns and rows before since inplace was set to false it is not showed.
```

```python
#Here is where inplace is set to true
df.drop('new', axis=1, inplace=True)
```

```python
df
```

```python
#As you can see, the new column has been completely dropped!
```

# DataFrames -- Part 3:

```python
# Index Levels
outside = ['G1','G1','G1','G2','G2','G2']
inside = [1,2,3,1,2,3]
hier_index = list(zip(outside,inside))
hier_index = pd.MultiIndex.from_tuples(hier_index)
#Multiindex?? You will see....
```

```python
df_2=pd.DataFrame(np.random.randn(6,2),index=hier_index,columns=['A','B'])
df_2 #Shows the multi index
```

```python
df_2
```

### Now Lets show how to index this? For the index hierarchy, we can use `df.loc`, and if this was on the column axis, you could just use normal bracket notation `df[]`. 

```python
df_2.loc['G1'] #Gives you the 'Subsection' under G1
```

```python
df_2.loc['G1'].loc[1] #You can keep on going deeper as well
```

```python
df_2.index.names
```

df.index.names

df.index.names is a property that returns the names of the row-index levels, not the individual row labels. It is accessed without parentheses.

A single-level index has one name; a MultiIndex has one name per level.

An unnamed level is represented by None.

When assigning names, the number of names must match the number of index levels.

python
df.index.names                         # View index-level names
df.index.names = ["Student"]           # Name a single-level index
df.index.names = ["Student", "Subject"] # Name an existing two-level index

In this data frame where it's a multi-index, you can do more than one for `df.index.names`, but in a normal data frame with just one index, it'll give you an issue if you try to do more than one. You just look at how many indexes you currently have. 

```python
#For example we can calll this outside index 'Groups and inside index 'num'
df_2.index.names = ['Group','Num']
```
