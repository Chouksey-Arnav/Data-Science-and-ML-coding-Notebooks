# Lecture number 3: Group By 

Explanation of Groupby: `GROUP BY` allows you to group together rows based on a column and perform an aggregate function on them. 

```python
import pandas as pd
import numpy as np
import matplotlib as plt
#import all out necessary libraries here
```

```python
data = {'Company':['GOOG','GOOG','MSFT','MSFT','FB','FB'],
       'Person':['Sam','Charlie','Amy','Vanessa','Carl','Sarah'],
       'Sales':[200,120,340,124,243,350]}
#Dictionary for data
```

```python
df=pd.DataFrame(data)
```

```python
df #Here is the OG, current loook of df that we got from the Dictionary we created!
```

```python
#Lets Groupby the Company column
byComp=df.groupby('Company') 
```

```python
#Then we are able to perform aggregate functions on what we have grouped by. 
```

```python
byComp['Sales'].mean() 
#As you can see the Group by, literally 'Groups By' Company so we get three different values. For the three different companies that we have grouped by here 
```

```python
byComp['Sales'].sum() #Group by comes in handy here as well
```

```python
byComp['Sales'].sum().loc['GOOG'] #We can also use row indexing here too!
```

```python
#But, we usually dont save the groupby in a variable like we have done here for explanatory purposes.
#Instead we can do it all in one line which is MUCH more efficient and easier once you get the hand of it!
df.groupby('Company')['Sales'].sum().loc['GOOG'] #We are able to get to the values MUCH quicker like this!
```

```python
df #Again this is the Database for better referance
```

```python
df.groupby('Company')['Sales'].max()
```

```python
df.groupby('Company').describe() #This function helps tell you a lot of the data for each company (in this case)!
```

```python
df.groupby('Company').describe().loc['MSFT'] #You can also look at specifics with .loc function for a specific company.
```
