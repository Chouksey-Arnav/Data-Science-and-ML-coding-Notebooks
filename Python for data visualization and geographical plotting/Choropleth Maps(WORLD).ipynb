```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
%matplotlib inline
```

```python
import plotly.express as px
import plotly.graph_objs as go
```

### Without further ado, and after importing all our necessary libraries, let's get started!

```python
df=pd.read_csv('2014_World_GDP')
df.head(2)
```

```python
data = go.Choropleth(
    locations=df['CODE'],
    z=df['GDP (BILLIONS)'],
    text=df['COUNTRY'],
    colorbar={'title': 'GDP in Billions'}
)
```

```python
layout = dict(
    title='GDP in Billions World Map (2014)',
    geo=dict(showframe=False, projection={'type': 'mercator'})
)
```

```python
maps = go.Figure(data=[data], layout=layout)
maps #World Map Variable
```
