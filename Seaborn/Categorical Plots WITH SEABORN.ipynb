```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
%matplotlib inline
```

```python
tips=sns.load_dataset('tips')# This is a preset dataframe
tips
```

```python
sns.barplot(x='sex',y='total_bill',data=tips) #) # What we did here is we set x as a categorical column and then y as a numerical column. In this case specifically, x was sex and y was total bill. 
```

### What this bar plot basically does is it shows us essentially based on groupings of sex, the average total bill. So we can see that male is a bit higher on average than female. Based on our tips dataset

### But what if we want to know like the standard deviation, like how much variance is here based on groupings of sex? We can add an extra estimator parameter after our specific dataset. After data, we can add a specific estimator parameter.
## Let me show you what I mean. 

```python
sns.barplot(x='sex',y='total_bill',data=tips, estimator=np.std) #The default estimator is set to the mean or average. 
```

### Now lets build off this through a countplot
### And it's basically what the count plot does is it counts the number of occurrences. So it counts how many times was the specific sex male versus female. It tells you that specific count based on our tips dataset that's predetermined. 

```python
sns.countplot(data=tips,x='sex') #) # Notice that you only pass in an x variable since you're only counting. 
#We can see there's more males in the dataset than females
```

### On a high level, count plot is the same as bar plot, but the y-axis has already been chosen for you and it's a specific count, and just to count the occurrences.

```python
tips #Here's our chosen DataFrame again. 
```

```python
sns.boxplot(data=tips,x='day',y='total_bill') # Notice that again, x is categorical and y is numerical. 
```

```python
sns.boxplot(data=tips,x='day',y='total_bill', hue='smoker') # Now you're also able to add this hue indicator as well. 
```

### Now let's talk about violin plots. Violin plots play a pretty similar role to box plots, and it's called through sns.violinplot.

```python
sns.violinplot(data=tips,x='day',y='total_bill') #) # Notice we'll just keep the same things for before, day as our category, and for our numerical value we'll do total bill so we can compare these 2 hand in hand. 
```

# STRIP PLOT:
### Strip plot draws a scatter plot where one variable is categorical, and we're going to do the exact same thing before where x is categorical, y is numerical, total bill, and x is going to be the day. So we're going to be comparing these 2, and then the data is obviously going to be equal to our tips DataFrame that we've already established and imported over from the Seaborn library and GitHub repository.

```python
sns.stripplot(data=tips,x='day',y='total_bill',jitter=True) #This is what you do for the stripplot it helps you see density.
```

### So the main point of a strip plot in Seaborn is to make sure all the clusters and training gaps in your data are all visible, since they can be hidden through summary charts like box plots. And they're also able to see, like, you can see where most of your data is concentrated. And also, if you have a category like days of the week, a strip plot lets you instantly compare how the numerical values spread out across each separate day.

# So all the plots we've been talking about so far, there's one thing that covers them all. So if you just use sns.catplot, you can just set the specific kind and then it can just create whatever you want for you. Let me show you what I mean by this: 

```python
sns.catplot(data=tips,x='day',y='total_bill',kind='bar')
```

### So what's really cool about this is you're able to switch the kind to anything. It's default set to the scatter plot, but you're able to switch the kind to anything. For example, you could set it to be bar, you can set it to be violin, and then you're able to do plots based on this in sns.catplot.
