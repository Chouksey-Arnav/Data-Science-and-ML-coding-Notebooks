### Okay, so the first type of plot that we're going to be talking about in Seaborn is called the distplot, which can be called through the function sns.displot. And what this really is, is you just pass in a single variable and it's able to create a good plot for you based on that.

```python
sns.displot(tips['total_bill'])
```

### So what this distplot function does is essentially just takes in one variable and it just gives us a histogram of the distribution of the total bill.

### And with this plot, you're able to make useful observations such as most of the bills are usually between $10 and $20, and then they start to fade away as it goes higher and higher and higher. So I guess there's less occurrences as you go higher and higher and higher.

## Now Lets go to JOINPLOTS in seaborne (sns.jointplot)

### Now in Seaborn, what a joint plot does is it allows you to basically match up the plots for bivariate data, meaning it's essentially 2 dis plots. So you're able to compare 2 columns instead of just one through a scatter plot. Let's take a look at how you can do that. By comparing total bill and tip columns in our dataset.

### You pass in an x variable or y variable for 2 things you want to compare, and the specific dataset that you're going to be using, in this case the tips dataset. 

```python
sns.jointplot(x='total_bill',y='tip',data=tips)
```

### Why this is helpful is because you're able to see a specific trend or relation between 2 different things. For example, in this case, the total bill column, the x-axis, and the tip column, which is the y-axis, shows that as the total bill goes up, the tip also goes up most of the time. So it helps us see that relation with this scatter plot.

```python
sns.pairplot(tips)
```

# RUG PLOT:
### Now let's explore a rug plot. All a rug plot is, is it's just a dash mark for every single point along the distribution line. So it's kind of like a distplot except, and you're only using comparing one specific column, but it's just a bit different way of looking at it.

```python
sns.rugplot(tips['total_bill'])
```

```python
#Here is our tips Dataframe again for better reference
tips.head(6)
```

```python
# Don't worry about understanding this code!
# It's just for the diagram below
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

#Create dataset
dataset = np.random.randn(25)

# Create another rugplot
sns.rugplot(dataset);

# Set up the x-axis for the plot
x_min = dataset.min() - 2
x_max = dataset.max() + 2

# 100 equally spaced points from x_min to x_max
x_axis = np.linspace(x_min,x_max,100)

# Set up the bandwidth, for info on this:
url = 'http://en.wikipedia.org/wiki/Kernel_density_estimation#Practical_estimation_of_the_bandwidth'

bandwidth = ((4*dataset.std()**5)/(3*len(dataset)))**.2


# Create an empty kernel list
kernel_list = []

# Plot each basis function
for data_point in dataset:
    
    # Create a kernel for each point and append to list
    kernel = stats.norm(data_point,bandwidth).pdf(x_axis)
    kernel_list.append(kernel)
    
    #Scale for plotting
    kernel = kernel / kernel.max()
    kernel = kernel * .4
    plt.plot(x_axis,kernel,color = 'grey',alpha=0.5)

plt.ylim(0,1)
```
