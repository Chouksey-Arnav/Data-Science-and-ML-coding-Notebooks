```python
import seaborn as sns
import matplotlib.pyplot as plt
iris_dataset=sns.load_dataset('iris')
```

```python
iris_dataset.head(1)
```

### This is a very famous dataset for measuring flowers, separating them into different columns such as sepal length, sepal width, petal length, petal width, and species. 

```python
sns.pairplot(data=iris_dataset) # if you forgot about this, revisit the distribution portion of this course!
```

```python
g=sns.PairGrid(data=iris_dataset)
g
```

### So what PairGrid essentially is, is that it creates an empty canvas. It's kind of like a pairplot, except you have a lot more control. 

```python
g = sns.PairGrid(iris_dataset)

g.map_diag(sns.histplot)
g.map_upper(sns.scatterplot)
g.map_lower(sns.kdeplot)
```

```python
#Now lets talk about the tips dataset
tips=sns.load_dataset('tips')
tips
```

```python
g=sns.FacetGrid(data=tips, col='time',row='smoker')
g.map(sns.histplot,'total_bill')
```
