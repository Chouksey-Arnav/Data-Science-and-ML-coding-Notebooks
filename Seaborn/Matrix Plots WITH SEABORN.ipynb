## Now in this notebook, we're going to be learning how to plot matrix plots, specifically heat maps, with the Seaborn library. Let's get started.

```python
 import seaborn as sns
```

```python
%matplotlib inline
```

```python
tips = sns.load_dataset('tips')
tips.head()
```

```python
penguins = sns.load_dataset("penguins")
penguins.head(6)
```

### So let's work it out with heatmaps. Now, in order for heatmaps to work, your data should already be in matrix form since that's what this entire notebook is about. It's about matrix plots with Seaborn. So what we're going to do, so we need it so that the index and the column match up. So cell value actually indicates something that is relevant to both of those names. What I mean by that is, for instance, if I take a look at the tips data frame right now, I have a total bill which is some sort of label or variable. And if I look at one of the values in the cell, it's $16.99, for example. However, the row is not an actual variable. In order to get this into matrix form, I need to have both variables on the columns and rows. So let's do that.

### So we will do this through the tips.corr method, CORR. 

```python
tc=tips.corr(numeric_only=True)
tc
```

###  So as you can see, now we have it in matrix form with both total bill, tip, and size on both the rows and columns. 

### So now we have it in the matrix form, so we can now call a heatmap on it. And it's very simple to call heatmap. All you have to do is...

```python
sns.heatmap(tc, annot=True, cmap='coolwarm') #As you can see here, we have changed the annotations that there's actual numbers inside, and we've changed the colors to fit the cool-warm palette.
```

### So now, as you can see here, we made a heatmap, but the other 2, the annotations equals to true and the cmap to whatever we set it as, in this example cool warm, what it does is first we can call heatmap on our specific Correlation data. So our correlation data matrix. For our heatmap, then we set annotation to true to tell us the exact numbers that the correlation is, and then we can set different tones with the color with the cmap.

```python
## Now let's go back and revisit the penguins data. 
```

```python
penguins
```

```python
#We need to get this penguins dataset now in a matrix form. 
```

```python
pen=penguins.pivot_table(index='species',columns='sex',values='body_mass_g')
pen
```

```python
sns.heatmap(data=pen,annot=True, cmap='coolwarm')
```

### Now let's talk about the second type of matrix plot in Seaborn, which is a cluster map. The cluster map is actually going to use hierarchical clustering to produce a clustered version of the heat map. Let me show you an example here where we have the original heat map, but now I'm going to call sns.cluster_map. Cluster map on our variable, and notice what we see here.

```python
sns.clustermap(data=pen,cmap='coolwarm')
```

### And as you can see now, things have been placed out of order as the cluster plot tries to cluster information. If we had a larger data frame instead of the penguins data frame, if you had a larger data set, now you can see this better, but just clusters things that are more similar together. It doesn't care about the preset order like the heatmap.

A normal heatmap will display things in order, and you can put them into cluster map to put things in order so that similar groups are closer to each other. As a summary.
