```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
%matplotlib inline
```

```python
tips=sns.load_dataset('tips')
tips.head(5)
```

### This is how we're going to make use of the tips dataset to make a very, very simple linear model plot with the `lmplot` function in seaborn. 


```python
sns.lmplot(x='total_bill',y='tip',data=tips,hue='sex') #Lm plot stands for 'linear model plot' 
```

### An 'lmplot' plot is very useful because, instead of just looking at a cloud of dots in a regular scatter plot, an ellipse plot automatically calculates and draws a linear regression line through your data. It even shades a light gray or colored bin around the line representing the confidence interval, which is how confident the model is about that trend.

### If you have categories in your data, like a hue, you can pass in a hue argument. `lmplot` will automatically color-code the dots and draw separate, independent regression lines for each category, allowing you to compare trends side by side. 

## What you're able to tell from this is that males and females have the same linear fit as far as total bill versus tip, so they're pretty similar. 

```python
sns.lmplot(x='total_bill',y='tip',data=tips,hue='sex', markers=['o','v']) 
#Now you can see that females are represented with the triangle marker, while males are represented with circle markers to see things better.
```
