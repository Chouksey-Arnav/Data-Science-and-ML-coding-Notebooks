```python
import matplotlib.pyplot as plt
```

```python
%matplotlib inline
```

```python
import numpy as np
```

```python
x=np.linspace(0,5,11)
x
```

```python
y=x**2
y
```

```python
plt.plot()#Creates a empty plot with plt.plot()
```

```python
plt.plot(x,y) #We plot thing with this
plt.show()
```

### Then you can also go a step forward and put a title, an x-label, and a y-label with matplotlib as well. 

```python
plt.plot(x, y)

plt.title('title')

plt.xlabel('x_label')

plt.ylabel('y_label') 
```

### Now, how do we put multiplots on a canvas? We're going to do this through the `plt.subplot` method in Matplotlib. 

```python
plt.subplot(1,2,1,)
plt.plot(x,y,'r')
plt.subplot(1,2,2)
plt.plot(y,x,'b')
```

### Okay, now the object-oriented method is here, so let's take a look at that. 

```python
fig=plt.figure()
axes=fig.add_axes([0.0,0.0,0.9,1])
axes.plot(x,y)
#We can set xlabels and ylabels too!
axes.set_ylabel('Y label')
axes.set_xlabel('X label')
axes.set_title('Title')
```

```python
fig2=plt.figure()
axes1=fig2.add_axes([1,1,1,1])
axes2=fig2.add_axes([0.3,0.4,0.6,0.8])
axes1.plot(x,y,'b')
axes1.set_title('axes1')
axes2.plot(y,x,'dr')
axes2.set_title('axes2')
```

## In matplotlib, what you do is you just create an empty figure, which acts like a blank canvas, and then you set axes for it.
