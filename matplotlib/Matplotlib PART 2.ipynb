```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
%matplotlib inline
```

```python
x=np.linspace(0,5,11)
x
```

```python
y=x**2
```

# Now that we have everything previously defined from the last notebook, we can go on to the new learnings.

###  okay, the `plt.subplots` function that we talked about in the last matplotlib part 1 notebook a tiny bit: now we're going to go more into it.

### The first two arguments that it takes in, if you press Shift-Tab, are `nrows` and `ncols`, which was previously automatically set to 1. That just gives you specific things replicated, and you're able to replicate into a matrix sort of system, but with graphs underneath them.

## Here's an example with us playing around with these specific features 

```python
fig,ax=plt.subplots(nrows=2,ncols=2)
```

```python
#Here's another one to solidify your understanding fully. 
fig,ax=plt.subplots(nrows=3,ncols=3)
#As you can see here, there are three rows of graphs and three rows of columns. 
```

### As you can see here as well, there's also lots of overlap. That's just a one-line fix. All we have to do is add `from matplotlib plt.tight_layout()`. 

```python
#So here are the lines from the last Jupyter cell exactly, word for word, and then, to stop that overlap from happening, all we have to do is add `plt.title_layout`. 
#Here's another one to solidify your understanding fully. 
fig,axes=plt.subplots(nrows=2,ncols=1)
#As you can see here, there are three rows of graphs and three rows of columns. 
plt.tight_layout()
```

```python
for ax in axes:
    ax.plot(x, y, 'b')
    ax.set_xlabel('x')
    ax.set_ylabel('y')
    ax.set_title('title')
plt.tight_layout()
# Display the figure object    
fig
 #Recommend it to call this at the end of every single time you plot. 
```

### Now let's talk about figure size and DPI. 

```python
fig=plt.figure(figsize=(8,2)) #for fig size, it measures how wide to make the specific graph, and the second number measures how tall to make it. You can play around with it here. 
```

### The same arguments can also be passed into layout managers such as the `subplots` function. 

```python
fig,axes=plt.subplots(figsize=(10,7))
axes.plot(x, y, 'b')
axes.set_xlabel('x')
axes.set_ylabel('y')
axes.set_title('title')
plt.tight_layout()
#We are able to change the figure sizes with this as well, with the `plt.subplots` function in Matplotlib
```

```python
#Now lets use other parameters in the plt.subplots function like the previously established nrows and ncols
fig,axes=plt.subplots(2,1,figsize=(8,2))
axes[0].plot(x,y)
axes[0].set_title('First Graph')
axes[1].plot(y,x)
axes[1].set_title('Second Graph')
plt.tight_layout()
```

```python
#In a circumstance like this, though, we'd want a longer second parameter, so it could fit everything much better. 
#Now lets use other parameters in the plt.subplots function like the previously established nrows and ncols
fig,axes=plt.subplots(2,1,figsize=(8,10))
axes[0].plot(x,y)
axes[0].set_title('First Graph')
axes[1].plot(y,x)
axes[1].set_title('Second Graph')
plt.tight_layout()
#This fits much better compared to the last block. 
```
