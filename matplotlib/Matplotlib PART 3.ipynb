### Let's first talk about setting colors with Matplotlib. 

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd 
x=np.linspace(0,5,11)
x
y=x**2
```

```python
fig=plt.figure()
#Lets add axes to this figure
ax=fig.add_axes([0,0,1,1])
ax.plot(x,y,color='#FF8C00') #You can put common colors into this, like blue, green, yellow, or orange, or you're able to put in a custom RGB hex code. 
```

### If you want a custom color for your specific line in matplotlib, you are able to search up through free tools, as there are lots of free tools online. Or you can just use Crayola colors, such as blue, green, yellow, purple, and pink. You get the idea for normal colors, but if you want custom colors for your line, that's also an option 

### Okay, that's the basics of colors. Now let's talk about line width and line style.  

```python
fig=plt.figure()
#Lets add axes to this figure
ax=fig.add_axes([0,0,1,1])
ax.plot(x,y,color='#FF8C00', linewidth=5, alpha=0.7, linestyle='--') #1 is default linewidth, and you are able to type in, for shorthand either lw or linewidth
#You can put common colors into this, like blue, green, yellow, or orange, or you're able to put in a custom RGB hex code. 
#Then, for transparency, in a line, you can add `alpha`. 1 is the default for `alpha` in Matplotlib, but you're able to change that based on how transparent or non-transparent you want the line. 
```

```python
fig=plt.figure()
#Lets add axes to this figure
ax=fig.add_axes([0,0,1,1])
ax.plot(x,y,color='#FF8C00', linewidth=5, alpha=0.7, linestyle='-', marker='o',markersize=10) #1 is default linewidth, and you are able to type in, for shorthand either lw or linewidth
#You can put common colors into this, like blue, green, yellow, or orange, or you're able to put in a custom RGB hex code. 
#Then, for transparency, in a line, you can add `alpha`. 1 is the default for `alpha` in Matplotlib, but you're able to change that based on how transparent or non-transparent you want the line. 
```

### You can also add an additional marker argument, and then you're able to mark down the specific points on the graph 

### Now, going past `marker`, you're also able to add other arguments. Past `marker`, you're able to add `marker`. Obviously, you're able to add `marker size`, which is how big you want the marker to be. You're able to add the `marker face color` argument and set it to a specific color if you want. You're also able to add the `marker edge width`. I'm going to edit that as well. Make that a little thicker, such as 3. That makes it a little thicker.

### Let's actually specify the color of the marker edge. I can specify a marker edge color, and let's go ahead and make that green. For the marker edge, you can also add the `marker color`, so you're able to specify the color for that 

```python
fig=plt.figure()
#Lets add axes to this figure
ax=fig.add_axes([0,0,1,1])
ax.plot(x,y,color='#FF8C00', linewidth=5, alpha=0.7, linestyle='-', marker='o',markersize=10, markerfacecolor='blue', markeredgewidth=2,markeredgecolor='purple') #1 is default linewidth, and you are able to type in, for shorthand either lw or linewidth
#You can put common colors into this, like blue, green, yellow, or orange, or you're able to put in a custom RGB hex code. 
#Then, for transparency, in a line, you can add `alpha`. 1 is the default for `alpha` in Matplotlib, but you're able to change that based on how transparent or non-transparent you want the line. 
```

```python
fig=plt.figure()
axes=fig.add_axes([0,0,1,1])
axes.plot(x,y,color='blue',linewidth=2,linestyle='--')
#Now let's say we only want this graph from 0 to 2 on the x-axis, so let's just say we want to limit it to that. 
axes.set_xlim(0,2)
axes.set_ylim(0,5)
```

### Essentially, like this, we are able to zoom in more on a specific matplotlib graph by setting our x-lim and y-lim through the `axes.set_xlim()` and the `axes.set_ylim()` And then, for the list, you pass in the upper bound and then the lower bound for both of them.
